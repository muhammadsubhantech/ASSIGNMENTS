## Q41. Student Performance System

```python
students = [
    {"name": "Alice", "scores": [85, 90, 78]},
    {"name": "Bob", "scores": [70, 75, 80]},
    {"name": "Charlie", "scores": [95, 88, 92]}
]
for student in students:
    total = 0
    for score in student["scores"]:
        total = total + score
    avg = total / len(student["scores"])
    print(student["name"], avg)
```

## Q43. Grade Book

```python
gradebook = {
    "Math": {"Alice": 85, "Bob": 70, "Charlie": 95},
    "Science": {"Alice": 90, "Bob": 75, "Charlie": 88},
    "English": {"Alice": 78, "Bob": 80, "Charlie": 92}
}
for subject in gradebook:
    scores = gradebook[subject]
    total = 0
    count = 0
    for name in scores:
        total = total + scores[name]
        count = count + 1
    print(subject, total / count)
```

## Q44. Course Enrollment

```python
courses = {
    "CS101": {"title": "Intro to CS", "students": ["Alice", "Bob", "Charlie"]},
    "MATH201": {"title": "Calculus", "students": ["Bob", "Diana", "Eve"]},
    "PHYS101": {"title": "Physics", "students": ["Alice", "Charlie", "Eve"]}
}
all_students = set()
for course in courses:
    for student in courses[course]["students"]:
        all_students.add(student)
print(all_students)
print(len(all_students))
```

## Q45. Sales Data Analysis

```python
sales = [
    {"product": "Laptop", "region": "North", "amount": 1500},
    {"product": "Mouse", "region": "South", "amount": 50},
    {"product": "Laptop", "region": "South", "amount": 1200},
    {"product": "Keyboard", "region": "North", "amount": 100},
    {"product": "Mouse", "region": "North", "amount": 75}
]
region_totals = {}
for sale in sales:
    region = sale["region"]
    amount = sale["amount"]
    if region in region_totals:
        region_totals[region] = region_totals[region] + amount
    else:
        region_totals[region] = amount
print(region_totals)
```

## Q46. Social Media Analytics

```python
posts = [
    {"user": "alice", "likes": 120, "tags": ["python", "coding"]},
    {"user": "bob", "likes": 80, "tags": ["data", "python"]},
    {"user": "charlie", "likes": 200, "tags": ["ai", "ml"]},
    {"user": "alice", "likes": 150, "tags": ["coding", "ai"]}
]
user_likes = {}
all_tags = set()
for post in posts:
    user = post["user"]
    likes = post["likes"]
    if user in user_likes:
        user_likes[user] = user_likes[user] + likes
    else:
        user_likes[user] = likes
    for tag in post["tags"]:
        all_tags.add(tag)
print(user_likes)
print(all_tags)
```

## Q48. E-commerce Order Processing

```python
orders = [
    {"order_id": "O001", "items": [{"name": "Laptop", "qty": 1, "price": 999}, {"name": "Mouse", "qty": 2, "price": 25}], "status": "pending"},
    {"order_id": "O002", "items": [{"name": "Keyboard", "qty": 1, "price": 45}], "status": "shipped"}
]
for order in orders:
    total = 0
    for item in order["items"]:
        total = total + item["qty"] * item["price"]
    print(order["order_id"], total, order["status"])
```

## Q49. Weather Data

```python
weather = {
    "Monday": {"temp": 22, "condition": "Sunny", "humidity": 60},
    "Tuesday": {"temp": 18, "condition": "Rainy", "humidity": 80},
    "Wednesday": {"temp": 25, "condition": "Sunny", "humidity": 55},
    "Thursday": {"temp": 20, "condition": "Cloudy", "humidity": 70},
    "Friday": {"temp": 23, "condition": "Sunny", "humidity": 50}
}
sunny_days = []
total_temp = 0
count = 0
for day in weather:
    total_temp = total_temp + weather[day]["temp"]
    count = count + 1
    if weather[day]["condition"] == "Sunny":
        sunny_days.append(day)
print(sunny_days)
print(total_temp / count)
```
