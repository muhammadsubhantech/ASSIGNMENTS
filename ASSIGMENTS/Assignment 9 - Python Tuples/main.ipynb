## Q11. Tuple Indexing

```python
colors = ("red", "green", "blue", "yellow", "purple")
print(colors[0])
print(colors[2])
print(colors[-1])
```

## Q12. Tuple Unpacking

```python
person = ("Alice", 25, "Engineer", "New York")
name = person[0]
age = person[1]
profession = person[2]
city = person[3]
print(name)
print(age)
print(profession)
print(city)
```

## Q13. Tuple Concatenation

```python
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)
combined = tup1 + tup2
new_tup = (combined[0], combined[-1])
print(combined)
print(new_tup)
```

## Q14. Count and Locate

```python
nums = (1, 2, 2, 3, 2, 4, 5, 2)
count = 0
first_index = -1
i = 0
while i < len(nums):
    if nums[i] == 2:
        count = count + 1
        if first_index == -1:
            first_index = i
    i = i + 1
print(count)
print(first_index)
```

## Q15. Extended Tuple Unpacking

```python
data = (10, 20, 30, 40, 50, 60)
first = data[0]
last = data[-1]
middle = data[1:-1]
print(first)
print(last)
print(middle)
```

## Q16. Modify an Immutable Tuple

```python
colors = ("red", "green", "blue")
new_colors = (colors[0], "yellow", colors[2])
print(new_colors)
```

## Q18. Tuple Sorting

```python
scores = (
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 78),
    ("Diana", 95)
)
sorted_scores = list(scores)
n = len(sorted_scores)
i = 0
while i < n:
    j = 0
    while j < n - 1:
        if sorted_scores[j][1] < sorted_scores[j + 1][1]:
            temp = sorted_scores[j]
            sorted_scores[j] = sorted_scores[j + 1]
            sorted_scores[j + 1] = temp
        j = j + 1
    i = i + 1
print(sorted_scores)
```

## Q19. Tuple Matrix

```python
matrix = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
middle_row_sum = matrix[1][0] + matrix[1][1] + matrix[1][2]
print(diagonal)
print(middle_row_sum)
```

## Q20. Inventory Tuple Challenge

```python
inventory = (
    ("Laptop", 5, 999.99),
    ("Mouse", 20, 25.50),
    ("Keyboard", 15, 45.00)
)
total_value = 0
i = 0
while i < len(inventory):
    item = inventory[i]
    value = item[1] * item[2]
    print(item[0], value)
    total_value = total_value + value
    i = i + 1
print(total_value)
```
