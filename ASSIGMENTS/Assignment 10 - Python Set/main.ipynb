## Q21. Unique Values

```python
nums = [10, 20, 10, 30, 20, 40, 10]
unique = set(nums)
print(unique)
```

## Q22. Set Operations

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
```

## Q23. Add and Remove

```python
colors = {"red", "green", "blue"}
colors.add("yellow")
colors.add("red")
colors.remove("green")
print(colors)
```

## Q24. Common Students

```python
math_students = {"Alice", "Bob", "Charlie", "David"}
science_students = {"Bob", "David", "Eve", "Frank"}
common = math_students & science_students
print(common)
```

## Q25. Subset and Superset

```python
required = {"math", "english", "science"}
student1 = {"math", "english", "science", "history"}
student2 = {"math", "english"}
print(required.issubset(student1))
print(required.issubset(student2))
print(student1.issuperset(required))
```

## Q26. Missing Numbers

```python
nums = {1, 2, 4, 6, 7, 9}
all_nums = set()
i = 1
while i <= 10:
    all_nums.add(i)
    i = i + 1
missing = all_nums - nums
print(missing)
```

## Q27. Three-Set Analysis

```python
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}
set3 = {5, 6, 7, 8, 9}
common_all = set1 & set2 & set3
only_set1 = set1 - set2 - set3
print(common_all)
print(only_set1)
```

## Q28. Transaction Analysis

```python
day1 = {"A123", "B456", "C789", "D012"}
day2 = {"B456", "C789", "E345", "F678"}
day3 = {"C789", "F678", "G901", "H234"}
all_customers = day1 | day2 | day3
repeat = (day1 & day2) | (day2 & day3) | (day1 & day3)
print(all_customers)
print(repeat)
```

## Q29. Access Control

```python
admins = {"alice", "bob", "charlie"}
editors = {"bob", "david", "eve"}
viewers = {"charlie", "eve", "frank", "grace"}
all_users = admins | editors | viewers
can_edit = admins | editors
print(all_users)
print(can_edit)
```

## Q30. Customer Segmentation

```python
premium = {"P001", "P002", "P003", "P004", "P005"}
active = {"P002", "P003", "P006", "P007", "P008"}
recent = {"P003", "P004", "P008", "P009", "P010"}
target = premium & active & recent
only_premium = premium - active - recent
print(target)
print(only_premium)
```
