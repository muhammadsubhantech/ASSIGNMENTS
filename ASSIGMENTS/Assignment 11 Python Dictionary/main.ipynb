## Q34. Merge Dictionaries

```python
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"c": 4, "d": 5, "e": 6}
merged = {}
for key in dict1:
    merged[key] = dict1[key]
for key in dict2:
    merged[key] = dict2[key]
print(merged)
```

## Q35. Student Marks

```python
marks = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95, "Eve": 88}
total = 0
count = 0
highest_name = None
highest_score = 0
for name in marks:
    score = marks[name]
    total = total + score
    count = count + 1
    if score > highest_score:
        highest_score = score
        highest_name = name
average = total / count
print(highest_name)
print(average)
```

## Q36. Nested Employee Records

```python
employees = {
    "E001": {"name": "Alice", "dept": "HR", "salary": 60000},
    "E002": {"name": "Bob", "dept": "IT", "salary": 75000},
    "E003": {"name": "Charlie", "dept": "IT", "salary": 80000}
}
print(employees["E002"]["name"])
print(employees["E003"]["salary"])
it_salaries = 0
it_count = 0
for emp_id in employees:
    if employees[emp_id]["dept"] == "IT":
        it_salaries = it_salaries + employees[emp_id]["salary"]
        it_count = it_count + 1
print(it_salaries / it_count)
```

## Q37. Dictionary Inversion

```python
original = {"a": 1, "b": 2, "c": 3, "d": 2}
inverted = {}
for key in original:
    value = original[key]
    if value not in inverted:
        inverted[value] = []
    inverted[value].append(key)
print(inverted)
```

## Q38. Order Calculation

```python
prices = {"apple": 1.5, "banana": 0.5, "orange": 1.2, "mango": 2.0}
order = {"apple": 3, "banana": 6, "orange": 2}
total = 0
for item in order:
    total = total + prices[item] * order[item]
print(total)
```

## Q39. Employee Performance

```python
performance = {
    "Alice": {"Q1": 80, "Q2": 85, "Q3": 90, "Q4": 88},
    "Bob": {"Q1": 70, "Q2": 75, "Q3": 80, "Q4": 85},
    "Charlie": {"Q1": 90, "Q2": 92, "Q3": 88, "Q4": 95}
}
for name in performance:
    scores = performance[name]
    total = scores["Q1"] + scores["Q2"] + scores["Q3"] + scores["Q4"]
    avg = total / 4
    print(name, avg)
```

## Q40. Inventory Management

```python
inventory = {
    "Laptop": {"quantity": 10, "price": 999.99},
    "Mouse": {"quantity": 50, "price": 25.50},
    "Keyboard": {"quantity": 30, "price": 45.00}
}
inventory["Laptop"]["quantity"] = inventory["Laptop"]["quantity"] - 2
inventory["Monitor"] = {"quantity": 15, "price": 299.99}
total_value = 0
for item in inventory:
    total_value = total_value + inventory[item]["quantity"] * inventory[item]["price"]
print(inventory)
print(total_value)
```
